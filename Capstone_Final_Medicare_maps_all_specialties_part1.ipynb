## Creating a dataframe of only New Mexico Providers

```
filter = (df['State'] == "NM")
df_nm = df[ filter ]
df_nm
```

```
distinct_count = df_nm['NPI'].nunique()
distinct_count
```

There are ~6000 duplicate providers so we will delete the duplicates that do not provide services to multiple zip codes.  Providers in New Mexico may provide services in more than one zip code.

```
# Standardize the ZIP column to 5 digits
df_nm['zip_5'] = df_nm['ZIP Code'].astype(str).str.extract(r'(\d+)', expand=False).str.zfill(5).str[:5]
```

```
df_nm
```

```
df_nm = df_nm.drop_duplicates(subset=['NPI', 'zip_5'])
```

```
df_nm = df_nm.drop(columns=[
    'NPI', 'Ind_PAC_ID', 'Ind_enrl_ID',
    'Provider Middle Name', 'suff', 'Med_sch',
    'Telephone Number', 'ind_assgn', 'grp_assgn',
    'adrs_id', 'Facility Name', 'org_pac_id',
    'num_org_mem', 'adr_ln_1', 'adr_ln_2', 'ln_2_sprs', 'Grd_yr',
    ])
df_nm
```

## Looking at the specialties of interest

This project is focused on geriatric needs.  Primary care is typically by Family Practice, who treat a broad range of ages, and Geriatric Medicine specialists, whose focus is on geriatric patients.  After speaking with our SMEs, Cardiology, Nephrology and Oncology were identified as the most critical specialties that affect geriatric patients.  Seperate dataframes were created for each of these groups of provides.  Additionally, as Physician Assistants and Nurse Practitioners provide care, we explored these providers as well.

The CMS data is broken down into 5 useful rows for specialties, a primary specialty and 4 rows of secondary specialties. To determine what specialties needed to be combined to create these tables was explored by looking for what specific language was used for each possible specialty row.

There is no secondary specialties recorded for either PAs or NPs which limits the utility of this data for evaluating practice beyond physicians

Exploring geriatric providers, this set is specific to the term "GERIATRIC MEDICINE' in any of the specialty columns.  There are providers that list GERIATRIC MEDICINE as a primary (pri_spec), secondary spec 1 and 2 and sec_spec_all.

```
df_nmger = df_nm.loc[(df_nm == 'GERIATRIC MEDICINE').any(axis=1)]
df_nmger
```

As family practice providers which are primary providers often treat geriatric patients, I pulled the providers listed as "FAMILY PRACTICE' in any of the specialty columns.  There are providers that list FAMILY PRACTICE as a primary (pri_spec), secondary spec 1 and 2 and sec_spec_all.

```
df_nmfam = df_nm.loc[(df_nm == 'FAMILY PRACTICE').any(axis=1)]
df_nmfam
```

Next, I created the specialist dataframes.  The cardiac specialties that are included are: CARDIOVASCULAR DISEASE (CARDIOLOGY), INTERVENTIONAL CARDIOLOGY, and CARDIAC ELECTROPHYSIOLOGY. As with the other searches, this included the primary specialty and all secondary specialties.

```
cardiac_specs = ['CARDIOVASCULAR DISEASE (CARDIOLOGY)', 'INTERVENTIONAL CARDIOLOGY', 'CARDIAC ELECTROPHYSIOLOGY']

df_nmcard = df_nm.loc[df_nm.isin(cardiac_specs).any(axis=1)]
df_nmcard
```

As we have a strong interest in nephrology, I grabbed the nephrology specialists in case we get a chance to explore these as well. This search is simple as there is only one term for NEPHROLOGY.

```
df_nmneph = df_nm.loc[(df_nm == 'NEPHROLOGY').any(axis=1)]
df_nmneph
```

Finally, I pulled oncology specialties in case we get a chance to explore a third interest. The specialties for oncology are: HEMATOLOGY/ONCOLOGY, SURGICAL ONCOLOGY, MEDICAL ONCOLOGY.

```
onco_specs = ['HEMATOLOGY/ONCOLOGY', 'SURGICAL ONCOLOGY', 'MEDICAL ONCOLOGY']

df_nmonco = df_nm.loc[df_nm.isin(onco_specs).any(axis=1)]
df_nmonco
```

## Automate geocoding and mapping

The zip codes will be used to get the latitude and longitude for each provider to allow interactive mapping.  Two methods of interactive mapping were used.  The first creates nicer maps, but can interact poorly in live display and with different browsers. However, when opened in at least Microsoft edge works well.  The second map style interacts more cleanly and reliably in Colab. Both types of maps are saved to files to give the viewer more options for utilizing the maps.     

```
%pip install pgeocode
```

```
import pgeocode

nomi = pgeocode.Nominatim('us')
```

Adding longitude and latitude to dataframes

### Family Practice

```
# Batch lookup coordinates using the series
geo_data = nomi.query_postal_code(df_nmfam['zip_5'].tolist())
geo_data
```

```
# Join coordinates and county names back to your dataframe
df_nmfam['latitude'] = geo_data['latitude'].values
df_nmfam['longitude'] = geo_data['longitude'].values
df_nmfam['county_name'] = geo_data['county_name'].values
```

```
# Drop any rows that couldn't be resolved
df_nmfam = df_nmfam.dropna(subset=['latitude', 'longitude'])

print(f"Successfully geocoded {len(df_nmfam)} providers via ZIP code centroids!")
```

```
import folium
from folium.plugins import MarkerCluster

# 1. Initialize map centered on New Mexico
nm_map = folium.Map(location=[34.5199, -105.8701], zoom_start=7)
```

```
# 3. Overlay the Cardiology Provider Markers using a Cluster layer for clean viewing
marker_cluster = MarkerCluster(name="Family Practice").add_to(nm_map)

for idx, row in df_nmfam.iterrows():
    popup_text = f"Dr. {row['Provider First Name']} {row['Provider Last Name']}"

    folium.Marker(
        location=[row['latitude'], row['longitude']],
        popup=popup_text,
        icon=folium.Icon(color='blue', icon='heart', prefix='fa')
    ).add_to(marker_cluster)

# 4. Add layer controls so you can toggle views on/off
folium.LayerControl().add_to(nm_map)

# Save or display the map
nm_map.save("nm_family_practice_risk_map.html")
nm_map
```

```
from folium.plugins import MarkerCluster
from IPython.display import HTML

# 1. Build the map using CartoDB tiles (doesn't trigger the OSM 403 block)
nm_map = folium.Map(location=[34.512, -105.988], zoom_start=7, tiles="CartoDB positron")
marker_cluster = MarkerCluster(name="Family Practice").add_to(nm_map)

# 2. Map your 60 rows
for idx, row in df_nmfam.iterrows():
    popup_text = f"Dr. {row['Provider First Name']} {row['Provider Last Name']}"

    folium.Marker(
        location=[row['latitude'], row['longitude']],
        popup=popup_text,
        icon=folium.Icon(color='blue', icon='info-sign')
    ).add_to(marker_cluster)

folium.LayerControl().add_to(nm_map)

# 3. Save it locally
output_file = "nm_family_practice_risk_map2.html"
nm_map.save(output_file)

# 4. THE OFFLINE WORKAROUND: Read the raw text of the file and force-inject it
with open(output_file, 'r', encoding='utf-8') as f:
    html_content = f.read()

# Display it inside the notebook by passing the raw HTML text string
HTML(f'<div style="width:100%; height:1000px;">{html_content}</div>')
```
